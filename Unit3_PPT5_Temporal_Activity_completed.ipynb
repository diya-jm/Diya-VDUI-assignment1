## Part B — The vendor's sampling proposal (5 min) 
The vendor wants to record **one sample every 25 hours**. Traffic has a strong **24-hour cycle**.

**B1.** Implement:
- `nyquist_max_interval(period_h)`: the **largest** sampling interval that avoids aliasing (Nyquist: $f_s > 2f$)
- `aliased_period(true_period, sample_interval)`: the **apparent period**, using aliased frequency $f_a = |f - f_s|$ for $f_s/2 < f < 3f_s/2$ (the case in this activity)

Then compute `apparent_period_25h` for the vendor's plan.

```python
def nyquist_max_interval(period_h):
    return period_h / 2          # f_s > 2f  <=>  sampling interval < period / 2

def aliased_period(true_period, sample_interval):
    f, fs = 1 / true_period, 1 / sample_interval
    fa = abs(f - fs)
    return 1 / fa

apparent_period_25h = aliased_period(24, 25)    # 600 h = 25 days

check_B(nyquist_max_interval, aliased_period, apparent_period_25h)
```

```python
# (given) What the vendor's data would look like for spring 2017
seg = tr["traffic_volume"].loc["2017-03-01":"2017-05-31"].asfreq("h").interpolate(limit=3)
fig, ax = plt.subplots(figsize=(14, 3.5))
ax.plot(seg.index, seg.values, c="lightgrey", lw=0.4, label="true hourly")
ax.plot(seg.index[::25], seg.values[::25], "-o", c="red", ms=3, label="vendor: every 25 h")
ax.set_title(f"Sampling every 25 h turns the 24-h cycle into a fake ~{apparent_period_25h/24:.0f}-day wave"); ax.legend(); plt.show()
```

**B2. A manager looks only at the vendor's 25-hour samples and says: *'Traffic rises and falls on a roughly 25-day cycle — let's schedule closures in the low phase.'* Explain what is really happening, using the words sampling rate, Nyquist and aliasing.**

 *Your answer:* The vendor's **sampling rate** (one sample per 25 h) is far below what a 24-h cycle needs, so the data is **aliased**. Each sample lands one hour earlier in the daily cycle than the previous one, so the samples drift slowly through the cycle and only come back to the same phase after 24 × 25 = 600 hours, which is 25 days. That slow wave is not in the traffic at all: it is an alias created by sampling below the **Nyquist** rate (which needs more than two samples per cycle, i.e. an interval under 12 h). Scheduling closures in its "low phase" would be acting on an artefact.

**B3. What maximum sampling interval would you recommend to MnDOT? Would sampling exactly every 12 hours be safe? Justify briefly.**

*Your answer:* The interval must be **shorter than 12 hours** for a 24-h cycle (Nyquist: f_s > 2f). Exactly 12 h is **not safe**, because the condition is strict: at exactly two samples per cycle the samples can fall on the same phase points every day (for example on the zero-crossings) and miss the cycle, and real traffic also has sharper features such as two rush-hour peaks (faster harmonics) that need a much shorter interval. I would recommend keeping the existing **hourly** sampling, or at the very least no more than a few hours.

**C2. In the zoom, the trailing average dips *after* 4 July while the centred one dips *around* it. How many days does a 7-day trailing average lag, and why does that matter to someone reading the chart?**

 *Your answer:* A 7-day trailing average at day t averages days t−6 to t, so it is centred 3 days in the past: it **lags by 3 days**. This matters because the dip caused by 4 July shows up on the chart about 3 days late, so a reader would attribute an effect to the wrong dates (and could think the holiday caused something after it). The centred average lines up with the event and is better for judging when things happened, while the trailing average is the one you can compute live, as data arrives.

**C3. The engineers care about which *day of the week* is quietest. Which is the right view: raw daily, 7-day MA, or 30-day MA? What does each larger window hide? Also interpret your `points_per_pixel` value.**

*Your answer:* For the day of the week, **raw daily** values are the right view. The `points_per_pixel` value (several hourly points competing for each pixel) shows that plotting raw hourly data on a 1000-pixel chart overplots points and hides detail, which is why aggregating to daily totals is needed before plotting.

## Part D — Filtering and events: choose the closure hour (4 min) *(slides 24, 29)*
Crews can only work in **daytime**, starting between **9:00 and 15:00**. Find the **quietest start hour on a normal weekday**, and compare it with holidays.

💻 **D1.** Compute:
- `weekday_profile`: mean traffic **by hour (0–23)** for **Mon–Fri non-holiday** rows
- `holiday_profile`: mean traffic by hour for rows where `holiday_day != "None"`
- `best_closure_hour`: the hour between **9 and 15** (inclusive) where `weekday_profile` is lowest

**D2. You *filtered* (kept only weekday non-holiday rows) and then *aggregated* (mean per hour). What does each step keep and what does it hide? Name one risk of relying only on the mean profile for the closure decision.**

 *Your answer:* **Filtering** keeps the comparable "normal" conditions (Mon–Fri, not a holiday) and hides weekends, holidays and any day-specific effects, so you only see the days relevant to the closure. **Aggregating** to a mean per hour keeps the typical level at each hour of the day and hides the variability: the spread between days, outliers, differences between seasons and weather. The risk is that the quietest hour **on average** may still be busy on many individual days (for example in winter or in bad weather), so a decision based only on the mean could cause a closure at a time that is often busy. Showing a spread (percentile band or small multiples by season) would reduce this risk.

## Part E — Design the animated *Traffic Pulse* for the lobby screen (6 min) 
**Design brief:** the screen animates **24 hourly states** of a typical weekday. Each hour is shown for **0.75 s** at **30 fps**, with a smooth transition between hours.

💻 **E1.** Implement and compute:
- `frame_budget_ms(fps)`: milliseconds available per frame
- `smoothstep(u)`: the ease-in-out function $3u^2-2u^3$
- `n_frames`, `duration_s`: for the brief above
- `compression`: seconds of **data** per second of **animation** (one hour of data per 0.75 s)

```python
def frame_budget_ms(fps):
    return 1000 / fps

def smoothstep(u):
    return 3 * u**2 - 2 * u**3

duration_s  = 24 * 0.75                 # 24 hourly states x 0.75 s = 18 s
n_frames    = int(duration_s * 30)      # 18 s x 30 fps = 540 frames
compression = 3600 / 0.75               # 3600 s of data per 0.75 s of animation = 4800
```

💻 **E2.** A second version shows how the **weekday profile changes month by month** in 2017. Build `fig` with `px.bar` using **`animation_frame="Month"`**, and **fix the y-axis** so bars stay comparable across frames (congruence).

```python
# (given) monthly weekday profiles for 2017
wk17 = wk.loc["2017"]
mp = wk17.groupby([wk17.index.month.rename("month"), wk17.index.hour.rename("hour")])["traffic_volume"].mean().rename("vehicles").reset_index()
mp["Month"] = mp["month"].map(lambda m: calendar.month_abbr[m])
MONTHS = [calendar.month_abbr[i] for i in range(1, 13)]
mp.head()
```

```python
fig = px.bar(mp, x="hour", y="vehicles",
             animation_frame="Month", category_orders={"Month": MONTHS}, range_y=[0, 7500],
             title="Traffic Pulse 2017 — weekday profile by month")
fig.show()

check_E(frame_budget_ms, smoothstep, n_frames, duration_s, compression, fig)
```

```python
# (given) The SAME data as small multiples — compare with the animation
fig2 = px.bar(mp, x="hour", y="vehicles", facet_col="Month", facet_col_wrap=6, category_orders={"Month": MONTHS}, height=420,
              title="Small multiples: all 12 months at once")
fig2.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1])); fig2.show()
```

**E3. Play the animation and try to answer: *In which month is the 17:00 peak lowest?* Then answer it with the small multiples. Which was faster and more certain, and why? Use the terms working memory and 'eyes beat memory'.**

*Your answer:* The **small multiples** were faster and more certain. In the animation, the 17:00 bar of each month is visible only for a moment, so I had to remember the height from earlier frames and compare it with the present one, which relies on **working memory** and is unreliable for 12 values. With small multiples all 12 months are on the screen at once and I can compare them directly by eye: **eyes beat memory**, because the comparison is done by perception on the page and not by recall.

**E4. What would go wrong if you left out `range_y` (letting each frame auto-scale)? Which Heer & Robertson principle does that violate?**

*Your answer:* Without a fixed `range_y`, every frame would **auto-scale** its y-axis to its own data, so a bar of the same height would mean different numbers in different months, and a month with low traffic would look just as tall as a busy one (the evening peak could even look unchanged). This breaks the **congruence principle** of Heer & Robertson (the structure and content of the animation should match the structure and content of the data being conveyed), because the visual change would not correspond to the real change.

**E5. Why use smoothstep (slow-in/slow-out) between hours instead of linear motion or a jump cut? Refer to your `smoothstep(0.25)` value and the per-frame jumps at the start and end.**

 *Your answer:* `smoothstep(0.25) = 0.156`: after a quarter of the transition only about 16% of the change has happened (linear motion would be at 25%), so the movement **starts and ends slowly and moves fastest in the middle**. With 22.5 frames per hour at 30 fps, the first frame moves only about 0.6% of the change (linear: 4.4%), and the same at the end, so there is no abrupt start or stop. This slow-in/slow-out motion looks natural and lets the eye track the bars, whereas linear motion looks mechanical and a jump cut gives no chance to see what changed and what stayed the same (no object constancy).

Write your recommendation to your manager.

**F1. For the **road-works engineers**: which visualization(s) should they get, and should they be animated? Name at least two idioms from the lecture and justify.**

 *Your answer:* The engineers should get **static, interactive** views and no animation: a **line chart of the hourly profile** (weekday vs holiday, with the allowed 9:00–15:00 window highlighted, as in Part D) and **small multiples** (for example one panel per month or season) so they can compare all months at once. **F2. For the **public lobby screen**: animation is acceptable here. Give three concrete design rules you'd apply (timing, fps, scales, labels, playback) so the public actually perceives the change.**

 *Your answer:* (1) **Timing and frame rate:** hold each state long enough to read (about 0.75 s per hour in this brief) and use 30 fps with eased (smoothstep) transitions, so the motion is smooth but not too fast to follow. (3) **Clear labels and playback control:** show the current hour or month large and clearly, annotate the key moment (for example the evening peak), and allow pause and replay or a scrubber so that viewers can look again at the moment they missed.

**F3. A viewer watching the lobby animation fails to notice that the evening peak disappears in one month. What is this phenomenon, and name two ways to reduce it.**

 *Your answer:* This is **change blindness**: when the changes between frames are small or happen while the viewer's attention is elsewhere (or at a moment of motion), people fail to notice them even if they are large. Ways to reduce it: **highlight or annotate** the change (an arrow, a colour flash, or a ghost outline of the previous frame), and **slow down or pause** at the important moment; showing the months side by side as **small multiples** (so the comparison does not depend on memory), or reducing the number of things that move at once, also helps.
